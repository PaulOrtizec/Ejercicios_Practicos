```python
import cv2
import urllib.request
import numpy as np
import matplotlib.pyplot as plt

# URL de la imagen externa
url = "https://i.pinimg.com/736x/7f/43/3f/7f433ff5238b59b0dd9b50ee69e9499d.jpg"  # Reemplaza con tu URL real

# Descargar la imagen desde la URL
resp = urllib.request.urlopen(url)
image_np = np.asarray(bytearray(resp.read()), dtype="uint8")
image = cv2.imdecode(image_np, cv2.IMREAD_COLOR)

# Verificar si se cargó bien
if image is None:
    print("❌ No se pudo cargar la imagen desde la URL.")
else:
    # Convertir a RGB para visualizar correctamente en matplotlib
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Crear la imagen más clara
    imagen_mas_clara = cv2.convertScaleAbs(image, alpha=1, beta=50)
    image_mas_clara_rgb = cv2.cvtColor(imagen_mas_clara, cv2.COLOR_BGR2RGB)

    # Mostrar ambas imágenes en una figura
    plt.figure(figsize=(12, 6))

    # Mostrar la imagen original
    plt.subplot(1, 2, 1)
    plt.imshow(image_rgb)
    plt.title("Imagen Original")
    plt.axis("off")

    # Mostrar la imagen con más brillo
    plt.subplot(1, 2, 2)
    plt.imshow(image_mas_clara_rgb)
    plt.title("Imagen con Más Brillo")
    plt.axis("off")

    # Mostrar ambas imágenes
    plt.show()
```

```python
import cv2
import urllib.request
import numpy as np
import matplotlib.pyplot as plt

# URL de la imagen externa
url = "https://i.pinimg.com/736x/7f/43/3f/7f433ff5238b59b0dd9b50ee69e9499d.jpg"  # Reemplaza con tu URL real

# Descargar la imagen desde la URL
resp = urllib.request.urlopen(url)
image_np = np.asarray(bytearray(resp.read()), dtype="uint8")
image = cv2.imdecode(image_np, cv2.IMREAD_COLOR)

# Verificar si se cargó bien
if image is None:
    print("❌ No se pudo cargar la imagen desde la URL.")
else:
    # Convertir a RGB para visualizar correctamente en matplotlib
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Crear la imagen más oscura (restando valor a los píxeles)
    imagen_mas_oscura = cv2.convertScaleAbs(image, alpha=1, beta=-50)  # Usando un valor negativo para oscurecer
    image_mas_oscura_rgb = cv2.cvtColor(imagen_mas_oscura, cv2.COLOR_BGR2RGB)

    # Mostrar ambas imágenes en una figura
    plt.figure(figsize=(12, 6))

    # Mostrar la imagen original
    plt.subplot(1, 2, 1)
    plt.imshow(image_rgb)
    plt.title("Imagen Original")
    plt.axis("off")

    # Mostrar la imagen más oscura
    plt.subplot(1, 2, 2)
    plt.imshow(image_mas_oscura_rgb)
    plt.title("Imagen con Más Oscuridad")
    plt.axis("off")

    # Mostrar ambas imágenes
    plt.show()
```

```python
import cv2
import urllib.request
import numpy as np
import matplotlib.pyplot as plt

# URL de la imagen externa
url = "https://i.pinimg.com/736x/7f/43/3f/7f433ff5238b59b0dd9b50ee69e9499d.jpg"  # Reemplaza con tu URL real

# Descargar la imagen desde la URL
resp = urllib.request.urlopen(url)
image_np = np.asarray(bytearray(resp.read()), dtype="uint8")
image = cv2.imdecode(image_np, cv2.IMREAD_COLOR)

# Verificar si se cargó bien
if image is None:
    print("❌ No se pudo cargar la imagen desde la URL.")
else:
    # Convertir a RGB para visualizar correctamente en matplotlib
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Crear la imagen más brillante
    imagen_mas_clara = cv2.convertScaleAbs(image, alpha=1, beta=50)
    image_mas_clara_rgb = cv2.cvtColor(imagen_mas_clara, cv2.COLOR_BGR2RGB)

    # Crear la imagen más oscura (restando valor a los píxeles)
    imagen_mas_oscura = cv2.convertScaleAbs(image, alpha=1, beta=-50)  # Usando un valor negativo para oscurecer
    image_mas_oscura_rgb = cv2.cvtColor(imagen_mas_oscura, cv2.COLOR_BGR2RGB)

    # Mostrar las tres imágenes en una figura
    plt.figure(figsize=(18, 6))

    # Mostrar la imagen original
    plt.subplot(1, 3, 1)
    plt.imshow(image_rgb)
    plt.title("Imagen Original")
    plt.axis("off")

    # Mostrar la imagen con más brillo
    plt.subplot(1, 3, 2)
    plt.imshow(image_mas_clara_rgb)
    plt.title("Imagen con Más Brillo")
    plt.axis("off")

    # Mostrar la imagen más oscura
    plt.subplot(1, 3, 3)
    plt.imshow(image_mas_oscura_rgb)
    plt.title("Imagen con Más Oscuridad")
    plt.axis("off")

    # Mostrar todas las imágenes
    plt.show()
```
